```python

# Imports
import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option("display.max_columns", 100)
data_path = Path("nashville_housing_data.csv")
assert data_path.exists(), "Place 'nashville_housing_data.csv' in the same folder as this notebook."
```

## 1) Load the dataset & basic profile

```python

# 1. Load the dataset
df = pd.read_csv(data_path)

# Display the first 5 rows
df.head()
```

```python

# Use .info() to inspect dtypes and missingness
df.info()
```

```python

# Numeric summary
df.describe(include=[np.number]).T
```

## 2) Standardize headings to lower_snake_case

```python

# Helper to convert column names to snake_case
import re

def to_snake(name: str) -> str:
    # Strip, lower, replace non-alphanumerics with underscores, collapse repeats
    s = name.strip().lower()
    s = re.sub(r"[^0-9a-z]+", "_", s)
    s = re.sub(r"_+", "_", s)
    return s.strip("_")

df.columns = [to_snake(c) for c in df.columns]
df.columns
```

## 3) Eliminate unneeded columns (`unnamed_0.1`, `unnamed_0`)

```python

drop_cols = [c for c in ["unnamed_0_1", "unnamed_0.1", "unnamed_0"] if c in df.columns]
# Note: If the original file had 'Unnamed: 0.1', after snake_case it may become 'unnamed_0_1'
df = df.drop(columns=drop_cols, errors="ignore")
drop_cols
```


## 4) Handle missing values
- Drop rows where `sale_price` is missing  
- Fill missing `bedrooms` with the median


```python

# Ensure sale_price is numeric before checking missingness
if "sale_price" in df.columns:
    df["sale_price"] = pd.to_numeric(df["sale_price"], errors="coerce")
else:
    raise KeyError("Expected column 'sale_price' not found after renaming.")

before_rows = len(df)
df = df.dropna(subset=["sale_price"])
after_rows = len(df)
dropped_rows = before_rows - after_rows

dropped_rows, after_rows
```

```python

# Fill missing bedrooms with median (coerce to numeric first)
if "bedrooms" in df.columns:
    df["bedrooms"] = pd.to_numeric(df["bedrooms"], errors="coerce")
    median_bed = df["bedrooms"].median()
    df["bedrooms"] = df["bedrooms"].fillna(median_bed)
else:
    print("Warning: 'bedrooms' column not found; skipping median imputation.")
median_bed if "median_bed" in locals() else None
```

## 5) Convert `sale_date` to datetime

```python

# Convert sale_date to datetime
if "sale_date" in df.columns:
    df["sale_date"] = pd.to_datetime(df["sale_date"], errors="coerce")
    df["sale_year"] = df["sale_date"].dt.year  # convenience feature
    df["sale_month"] = df["sale_date"].dt.month
else:
    print("Warning: 'sale_date' column not found; skipping datetime conversion.")
df.dtypes
```


## 6) Standardize `sold_as_vacant` to only **Yes** / **No**
Accept common variants like Y/Yes/True/1 and N/No/False/0.


```python

def normalize_yes_no(x):
    if pd.isna(x):
        return x
    s = str(x).strip().casefold()
    yes_set = {"y", "yes", "true", "1"}
    no_set  = {"n", "no", "false", "0"}
    if s in yes_set:
        return "Yes"
    if s in no_set:
        return "No"
    # If unknown value, leave as-is or set to NaN; here we leave as-is to avoid data loss
    return "Yes" if s.startswith("y") else ("No" if s.startswith("n") else np.nan)

if "sold_as_vacant" in df.columns:
    df["sold_as_vacant"] = df["sold_as_vacant"].apply(normalize_yes_no)
    # If any remaining NaN due to odd values, choose a default if needed; here we leave NaN.
else:
    print("Warning: 'sold_as_vacant' column not found; skipping standardization.")

df["sold_as_vacant"].value_counts(dropna=False) if "sold_as_vacant" in df.columns else "sold_as_vacant not present"
```

## 7) Save the cleaned dataset

```python

out_path = Path("nashville_housing_cleaned.csv")
df.to_csv(out_path, index=False)
out_path, df.shape
```
